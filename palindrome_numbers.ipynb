```python
def next_palindrome_bf(n):
    n += 1
    while (s:=str(n)) != s[::-1]:
        n += 1
    return n
```

```python
def next_palindrome(n):
    l,m,r = ((ns[:(sl:=len(ns)//2)], ns[sl:-sl], ns[-sl:])
             if len(ns:=str(n))>1
             else ('', str(n), ''))
    lp1 = str(int(l or '0')+1)
    return (int(f'{l}{m}{l[::-1]}') if r < l[::-1]                                 # If RHS < LHS, LHS+M+rev(LHS)
            else int(f"{lp1}{'0'*len(m)}"[:-1]+lp1[::-1]) if {*f'{l}{m}'}=={'9'}   # If LHS & M all 9s, (LHS+1)+(0 if M)+\b+rev(LHS+1)
            else int(f'{lp1}0{lp1[::-1]}') if m=='9'                               # If M==9, (LHS+1)+0+rev(LHS+1)
            else int(f'{l}{int(m)+1}{l[::-1]}') if len(m)                          # If len(M), LHS+(M+1)+rev(LHS)
            else int(f'{lp1}{lp1[::-1]}'))                                         # Else: (LHS+1)+rev(LHS+1)
```

```python
palindromes_v1 = [0]
for i in range(10000):
    palindromes_v1 += [next_palindrome_bf(palindromes_v1[-1])]
```

```python
palindromes_v2 = [0]
for i in range(1000000):
    palindromes_v2 += [next_palindrome(palindromes_v2[-1])]
```
