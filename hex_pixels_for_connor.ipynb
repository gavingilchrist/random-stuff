```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
grid = [[['3C65BC','70A0CD'][((i+6)//9+(j+6)//9)%2] for j in range(96)] for i in range(96)]
for i in range(11):
    for j in range(11):
        for ii in range(i*9+2,i*9+4):
            for jj in range(j*9+2,j*9+4):
                grid[ii][jj] = {'W':'FFFFFF','B':'000000'}['WBWWBWBB'[(j-i)%8]]

with open('data/nerd_stuff.txt', 'w') as f:
    f.write(' '.join(sum(grid,[])))
```

```python
plt.figure(figsize=(12.8, 9.6))
plt.imshow(np.vectorize(lambda x:int(x,16))(np.array(grid).view('<U2').reshape(96,96,3)))
```

```python
import matplotlib.image as mpimg
img = mpimg.imread('data/luna_nose.jpeg')

s=2
img_small = np.round(np.mean(np.stack([img[i:img.shape[0]//s*s:s,j:img.shape[1]//s*s:s]
                                       for i in range(s)for j in range(s)]),
                             axis=0),
                     0).astype(int)

hex_pixels = np.vectorize(lambda x:('00'+hex(x)[2:].upper())[-2:])(img_small).view('<U6').reshape(-1).tolist()
with open('data/wee_dug_big_nose.txt', 'w') as f:
    f.write(' '.join(hex_pixels))
```
