```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
```

```python
rgb_hex2dec = lambda x: [int(x[i:i+2],16) for i in range(0,len(x),2)]
rgb_hex2dec('014A50')
```

```python
# Lines
img = ((1-np.divide([[[87, 51, 51]]], 255))
       * np.repeat(np.expand_dims((np.arange(50)+1)//4%2 == 0, (1,2)), 1680, axis=1))
hfade = np.expand_dims(np.minimum(np.maximum(np.linspace(-0.25, 1.25, 1680), 0), 1), (0,2))
img = 1 - (img * hfade)

mpimg.imsave('assets/navbar.png', img)
```

```python
# Maze
rawmaze = mpimg.imread('assets/rawmaze.png')[:,:,0]
rawmaze = np.append(rawmaze[:-2], rawmaze, axis=0)
img = ((1-np.divide([[[87, 51, 51]]], 255))
       * (1-rawmaze[np.expand_dims(np.repeat(np.arange(0, rawmaze.shape[0], 8), 4), (0,2)),
                    np.expand_dims(np.repeat(np.arange(0, rawmaze.shape[1], 8), 4), (1,2))]))[1:-1, 2:-2]
hfade = np.expand_dims(np.minimum(np.maximum(np.linspace(-0.25, 2.6, img.shape[1]), 0), 1), (0,2))
img = 1 - (img * hfade)

mpimg.imsave('assets/navbar.png', img)
```
