```python
import numpy as np

# Import dictionary, make list of 4-letter words
with open('../data/word_list.txt', 'r') as f:
    words_4l = {i.upper() for i in f.read().split('\n') if len(i)==4}
```

```python
solutions = {'POOP': ('POOP', 0)}
wds = {'POOP'}
n_unsolved = len(words_4l)
while n_unsolved > (n_unsolved:=len(words_4l-{*solutions})):
    sol_new = {n: (f'{n}-{solutions[w][0]}', 1+solutions[w][1])
               for w in wds
               for l in [0,1,2,3]
               for r in {chr(i) for i in range(65,91)}-{w[l]}
               if (n:=f'{w[:l]}{r}{w[l+1:]}') in words_4l 
               and n not in solutions}
    solutions.update(sol_new)
    wds = {*sol_new}
solutions.update({i: 'NO SOLUTION' for i in words_4l-{*solutions}})
```
